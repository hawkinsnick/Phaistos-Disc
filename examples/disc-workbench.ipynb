# Disc workbench: reproducible descriptions
Run from the repository root or the examples folder. Native readings remain unchanged. All scenarios are conditional, and one object is not an independent sample of texts.

```python
import json, pathlib, collections
root = pathlib.Path.cwd()
if not (root / "analysis/edition-workbench-v1.json").exists(): root = root.parent
result = json.loads((root / "analysis/edition-workbench-v1.json").read_text())
assert result["scenario_count"] == 368
result["sensitivity_bounds"]
```

```python
native = next(s for s in result["scenarios"] if s["unknown_sign_number"] is None and s["b3_sign_number"] == 7 and not s["reverse_group_order"] and not s["reverse_within_group"])
native["all_slots"], native["graphical_groups"]
```

```python
corpus = json.loads((root / "corpus/disc.json").read_text())
values = {o["occurrence_id"]: o["sign_id"] for o in corpus["occurrences"]}
complete = [tuple(values[k] for k in g["occurrence_ids"]) for g in corpus["groups"] if all(values[k] is not None for k in g["occurrence_ids"])]
pair_count = sum(complete[i] == complete[j] for i in range(len(complete)) for j in range(i+1, len(complete)))
assert pair_count == 9
pair_count
```

No p-values, language identification or preferred reading follow from these descriptions. The full matrix retains source locators and explicit review limitations.
